## MSD vs Others

```python
df.shape
```

```python
df = df[df['wides'].isnull()]
df.shape
```

```python
df = df[df['season'] > 2019]
df.shape
```

```python
df_runs = pd.DataFrame(df.groupby('striker')['runs_off_bat'].sum()).reset_index()

df_balls = pd.DataFrame(df.groupby('striker')['ball'].count()).reset_index()
```

```python
df_all = df_runs.merge(df_balls,on = 'striker', how='left')
print(df_all.shape)
df_all.head()
```

```python
msd_df2['runs_off_bat'].sum()
```

```python
msd_df2['ball'].count()
```

```python
df_all= df_all[(df_all['runs_off_bat'] >= 500) & (df_all['ball'] <= 500)]
print(df_all.shape)
df_all.head()
```

```python
df_all['strike_rate']= df_all['runs_off_bat']/df_all['ball']*100
```

```python
df_all.head()
```

```python
df_all[df_all['striker'] == 'MS Dhoni']
```

for i in range(len(df_all)):
     plt.text(df_all['strike_rate'][i], df_all['runs_off_bat'][i], df_all['striker'][i])

```python
plt.figure(figsize=(18,9))
plt.scatter(df_all.strike_rate, df_all.runs_off_bat)

for strike_rate, runs_off_bat, striker in zip(df_all['strike_rate'], df_all['runs_off_bat'], df_all['striker']):
    plt.text(strike_rate, runs_off_bat +2, striker)
        
plt.title('Best Finishers Since IPL 2020-2022')
plt.xlabel('Strike Rate')
plt.ylabel('Runs')
plt.axvline(135, ls='--', color='black')
plt.axhline(535, ls='--', color='black')
plt.show()
```

```python
df_all.groupby(['striker'])['strike_rate'].sum().sort_values(ascending = False).plot(kind ='barh')
plt.xlabel('strike_rate')
plt.show()
```
